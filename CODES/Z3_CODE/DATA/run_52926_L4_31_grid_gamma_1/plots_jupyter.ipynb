```python
import numpy as np
import matplotlib.pyplot as plt
```

Plotting $A_\infty$

```python
time_list = np.exp(np.arange(-10, 25, 0.5))
data = data = np.load('./b110/autocorr_gamma_1_L4_f_1.11_phi_0.999_NPY.npy')[1]

target = 1/np.e

# Find index just after crossing
idx = np.argmax(data < target)

# Two points straddling the threshold
t0, t1 = time_list[idx-1], time_list[idx]
d0, d1 = data[idx-1], data[idx]

# Linear interpolation to find exact crossing time
tau = t0 + (target - d0) * (t1 - t0) / (d1 - d0)
decay_rate = 1 / tau

print(tau)

data_2 = data[0] * np.exp(-1 * decay_rate * time_list)

plt.plot(time_list, data, label='Autocorrelation')
plt.xscale('log')
plt.plot(time_list, data_2, label='Exponential Fit')
plt.legend()
plt.show()
```
