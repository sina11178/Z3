```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
N = 31  # GRID DIMENSION
x_vals = np.linspace(0, 1, N)  # x from 0 to 1.1
y_vals = np.linspace(0, 1, N)  # y from 0 to 1.1

# Initialize matrix: shape (31, 31, 2)~F~R each cell holds [r, phi]
grid = np.zeros((N, N, 2))

for i in range(N):       # row index �~F~R y-axis (bottom to top)
    for j in range(N):   # col index �~F~R x-axis (left to right)
        x = x_vals[j]
        y = y_vals[i]
        r   = np.sqrt(x**2 + y**2)
        phi = np.arctan2(y, x)   # in [0, pi/2] since x,y �~I� 0
        grid[i, j] = [r, phi]


gamma_array = [1]
L_array = [4,6,8]

time_list = np.exp(np.arange(-10, 25, 0.5))
```

```python
params = grid[15, 0]

L4_data = np.load(f'./L4/gamma_1/autocorr_gamma_1_L4_f_{round(params[0],3)}_phi_{round(params[1],3)}_NPY.npy')[1]
L6_data = np.load(f'./L6/gamma_1/autocorr_gamma_1_L6_f_{round(params[0],3)}_phi_{round(params[1],3)}_NPY.npy')[1]
L8_data = np.load(f'./L8/gamma_1/autocorr_gamma_1_L8_f_{round(params[0],3)}_phi_{round(params[1],3)}_NPY.npy')[1]

plt.plot(time_list, L4_data, label='L=4')
plt.plot(time_list, L6_data, label='L=6')
plt.plot(time_list, L8_data, label='L=8')

plt.xscale('log')
plt.legend()
plt.xlabel('Time')
plt.ylabel('Autocorrelation')
plt.title(f'Autocorrelation vs Time for f={round(params[0],3)}, phi={round(params[1],3)}')
plt.show()
```
